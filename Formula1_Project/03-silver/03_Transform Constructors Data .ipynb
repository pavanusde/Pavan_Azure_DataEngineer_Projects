# **Transform Constructors Data**
  1. Read bronze Constructors table
  2. Keep only the columns required for analytics (Drop url column)
  3. Standardise Column using snake_case (constructorid -> constructor_id)
  4. Rename Columns to make them more meaningful ( name -> constructor_name)
  5. Remove duplicate records 
  6. Transform values of columns nationality to Title Case 
  7. Write transformated data to constructors table 







```python
%run ../00-common/01_Environment-Config
```

```python
bronze_table_nm = f"{catalog_name}.{bronze_schema}.constructors"
silver_table_nm = f"{catalog_name}.{silver_schema}.constructors"
```

### Step 1 - Read bronze races table

```python
# #Below is one way of reading data from table 
# constructors_df = (
#     spark.table(bronze_table_nm)    
# )
```

```python
#Below is another way of reading data from table this one allows to add options 
constructors_df = (
    spark.read.table(bronze_table_nm)   
    
)
```

```python
display(constructors_df)
```

### Step 2 - Keep only the columns requried for analytics (Drop Url column)

```python
from pyspark.sql import functions as F
```

```python
constructors_select_df =  constructors_df.select(
        F.col("constructorId"),
        F.col("name"),        
        F.col("nationality"),          
        F.col("ingestion_timestamp"),
        F.col("source_file")
    )

    
```

### Step 3,4 - Standardise Column Names
Standardise Column using snake_case (constructorid -> constructor_id)

```python
constructors_renamed_df =  constructors_select_df.withColumnRenamed("constructorId", "constructor_Id")
```

```python
display(constructors_renamed_df)
```

### Step -5 Remove Duplicates Records

```python
 constructors_valid_df = constructors_renamed_df.dropDuplicates(["constructor_Id"])
 display(constructors_valid_df)
```

### Step - 6 Transform values of columns race_name to Title Case 

```python
constructors_final_df = constructors_valid_df.withColumns({
    "nationality": F.initcap(F.col("nationality"))
})
display(constructors_final_df)
```

### Step - 7 Write transformated data to silver races table 

```python
(
    constructors_final_df
        .write.mode("overwrite")
        .format("delta")
        .saveAsTable(silver_table_nm)
)
```
