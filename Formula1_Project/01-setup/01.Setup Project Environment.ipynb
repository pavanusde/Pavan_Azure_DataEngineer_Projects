# Setup Project Environment for Formula1 Project

### Create External Location 

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_ext_f1_dl1_dev
URL 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-f1-sc`)
COMMENT 'External Location For Formula1 Container'
```

### **Create Catalog Formula1**

```sql
CREATE CATALOG IF NOT EXISTS formula1
    MANAGED LOCATION 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net'
    COMMENT 'This is the main catalog for the Formula1 Project';
```

### **Create Schemas Landing , Bronze, Gold**

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net/bronze'
    COMMENT 'This is the Bronze schema for the Formula1 Project';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net/silver'
    COMMENT 'This is the Silver schema for the Formula1 Project';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net/gold'
    COMMENT 'This is the Gold schema for the Formula1 Project'        
```

```sql
USE CATALOG formula1;
SHOW SCHEMAS;
```

### **Create Volume Files**

```sql
CREATE EXTERNAL VOLUME formula1.landing.files 
LOCATION 'abfss://formula1@stf1devsouthus001.dfs.core.windows.net/landing';
```
