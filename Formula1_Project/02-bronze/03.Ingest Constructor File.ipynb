```python
source_file_name = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

#### **Step-1 Read the JSon file using the dataframe reader API**

```python
#Define The Schema 
constructors_schema = """constructorId STRING,
                        name STRING,
                        nationality STRING,
                        url STRING
                      """
```

```python
constructors_df = (
    spark.read
    .format("json")
#   .option("inferSchema", True)
    .schema(constructors_schema)
    .option("header", True)
    .option('mode','FAILFAST')
    .load(source_file_name)
)
```

#### **Step-2. Add Metadata Columns**

```python
constructors_final_df = add_ingestion_metadata(constructors_df)
```

```python
display(constructors_final_df)
```

####  **Step 3. Write to bronze delta table**

```python
(
    constructors_final_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(table_name)
)
```
