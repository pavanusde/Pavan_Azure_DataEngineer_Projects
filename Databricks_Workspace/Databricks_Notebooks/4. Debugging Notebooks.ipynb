### Demo 2
Demonstrate step in and step out

```python
def calculate_final_price(item_price, tax_rate, discount):
    #1.Calculate the tax to be applied 
    tax  = item_price * (tax_rate/100)
    #2.Apply  tax to the item price 
    item_price_with_tax = item_price + tax
    #3.Apply the flat discount to the item price 
    total = item_price_with_tax  - discount
    return total
```

```python
# Calculate the final price of an item and print the final total

item_price = 120 
tax_rate = 20 # Given Percentage 
discount = 10.00

# Calling calculate_final_price method and passing params
print(f"Total Final Price:{calculate_final_price(item_price, tax_rate, discount):.2f}")
```
